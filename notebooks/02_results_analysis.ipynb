# Results analysis
Compare the four ablation variants — Recall@k, MRR, NDCG@10, latency, token cost — and run paired significance tests.

## Baseline (PLABA)


```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Markdown

baseline = json.loads(Path("../results/plaba_baseline.json").read_text())
m = baseline["metrics"]
cfg = baseline["config"]
```

```python
Markdown(
    f"Recall@5 = **{m['recall_at_5']:.3f}** &middot; "
    f"Recall@10 = **{m['recall_at_10']:.3f}** &middot; "
    f"MRR = **{m['mrr']:.3f}** &middot; "
    f"NDCG@10 = **{m['ndcg_at_10']:.3f}** "
    f"(n={baseline['n_queries']}, chunks={baseline['n_chunks']}, "
    f"model={cfg['embedding_model']}, chunk={cfg['chunk_size']}/{cfg['chunk_overlap']})"
)
```

### Per-query distribution


```python
per_query = pd.DataFrame(baseline["per_query"])
per_query.describe()[["recall_at_5", "recall_at_10", "mrr", "ndcg_at_10", "latency_ms"]]
```
